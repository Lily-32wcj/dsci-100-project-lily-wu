#### Task 5.2 Perform the Analysis (20 points)

```r
library(tidymodels)
library(themis)
```

```r
count <- count(tidy_basketball, tourney_finish)
count #to check how imbalance the categories are
tidy_basketball2 <- tidy_basketball|>
filter(tourney_finish!="opening_round_loss")|>
droplevels() #remove opening_round_loss because it has too few observations to be meaningful
```

```r
 

basketball_split <- initial_split(tidy_basketball2, prop = 0.7, strata = tourney_finish)
basketball_train <- training(basketball_split)
basketball_test<-testing(basketball_split)   #split data into 70% training set and 30% testing set with equal proportions of each tourney_finish category
 
basketball_recipe <- recipe(tourney_finish ~., data = basketball_train) |>
step_upsample(tourney_finish, over_ratio=1)   #create a recipe and preprocess data, upsampling to balance tourney_finish categories 

basketball_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
                      set_engine("kknn") |>
                      set_mode("classification")  #build a knn classification model to choose k later

basketball_vfold <- vfold_cv(basketball_train, v = 3, strata = tourney_finish)  #split training data into 3 folds to train and evaluate k values later

basketball_workflow <- workflow() |>
                        add_recipe(basketball_recipe) |>
                        add_model(basketball_spec)  #create a workflow with recipe and model

gridvals <- tibble(neighbors = seq(from=10,to=100,by=5))  #decide values of k to test on: 10, 15, 20, 25... to 100.

k_max <- basketball_workflow |>
                        tune_grid(resamples = basketball_vfold, grid = gridvals) |>   #fit the model on each value of k
                        collect_metrics()|>     #collect performance metrics
                        filter(.metric=="accuracy")|>
                        filter(mean==max(mean))|>
                        pull(neighbors)     #output the k value with largest average accuracy across folds

k_max   #print k value

k_basketball_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = k_max) |>   #input the selected k value to the model
  set_engine("kknn") |>
  set_mode("classification")

basketball_fit <- workflow() |>
  add_recipe(basketball_recipe) |>
  add_model(k_basketball_spec) |>
  fit(data = basketball_train)    #fit and train the final model on training data

basketball_preds <- basketball_fit |>
  predict(basketball_test) |>   #predict the testing data using the trained model
  bind_cols(basketball_test)   
head(basketball_preds)  #preview some predictions compared to actual observations

basketball_evaluation <- basketball_preds |>
  metrics(truth = tourney_finish, estimate = .pred_class)|>
  filter(.metric == "accuracy") |>
  pull(.estimate)    #collect the performance metrics of testing data. Output the accuracy of predictions.
basketball_evaluation   #print accuracy of testing data predictions
```
